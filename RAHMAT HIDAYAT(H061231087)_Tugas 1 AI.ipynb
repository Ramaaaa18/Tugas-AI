5. Kewajaran interpretasi litologi berdasarkan nilai kecepatan gelombang P dan informasi geologi yang tersedia.



```python

# ============================================================
# ANALISIS DATA SEISMIK REFRAKSI - METODE INTERCEPT TIME
# Data simulasi: 24 geofon, 3 lapisan, forward dan reverse shot
# Satuan: jarak (m), waktu (s), kecepatan (m/s), kedalaman (m)
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import linregress

# ------------------------------------------------------------
# 1. PARAMETER DATA SIMULASI
# ------------------------------------------------------------

n_geofon = 24
spasi_geofon = 2.0

# Posisi geofon dalam satu lintasan
x_geofon = np.arange(n_geofon) * spasi_geofon
panjang_lintasan = x_geofon[-1]

# Parameter model simulasi (bukan hasil pengukuran lapangan)
V_model = np.array([500.0, 1000.0, 2500.0])
h_model = np.array([3.0, 4.0])  # ketebalan lapisan 1 dan 2

V1, V2, V3 = V_model
h1_model, h2_model = h_model

# ------------------------------------------------------------
# 2. MENGHITUNG WAKTU INTERSEP TEORITIS
# ------------------------------------------------------------

ti2_model = (
    2 * h1_model * np.sqrt(V2**2 - V1**2) / (V1 * V2)
)

ti3_model = (
    2 * h1_model * np.sqrt(V3**2 - V1**2) / (V1 * V3)
    + 2 * h2_model * np.sqrt(V3**2 - V2**2) / (V2 * V3)
)

print("WAKTU INTERSEP TEORITIS")
print(f"ti lapisan 2 = {ti2_model:.6f} s")
print(f"ti lapisan 3 = {ti3_model:.6f} s")

# ------------------------------------------------------------
# 3. MEMBUAT DATA FIRST ARRIVAL SIMULASI
# ------------------------------------------------------------

# Tiga cabang kurva waktu tempuh:
# cabang 0 = gelombang langsung pada lapisan 1
# cabang 1 = gelombang refraksi pada lapisan 2
# cabang 2 = gelombang refraksi pada lapisan 3

def hitung_cabang(x):
    return np.column_stack([
        x / V1,
        ti2_model + x / V2,
        ti3_model + x / V3
    ])

data = []

for arah in ["Forward", "Reverse"]:
    # Jarak dari sumber untuk masing-masing arah
    if arah == "Forward":
        jarak = x_geofon.copy()
    else:
        jarak = panjang_lintasan - x_geofon

    cabang = hitung_cabang(jarak)

    # First arrival adalah waktu tempuh terkecil
    indeks_cabang = np.argmin(cabang, axis=1)
    waktu_teori = np.min(cabang, axis=1)

    # Noise kecil untuk membuat simulasi lebih realistis
    # Seed tetap agar hasil selalu sama saat dijalankan
    rng = np.random.default_rng(
        42 if arah == "Forward" else 43
    )
    noise = rng.normal(0, 0.00008, size=n_geofon)

    waktu_observasi = np.maximum(waktu_teori + noise, 0)

    for i in range(n_geofon):
        data.append({
            "arah_shot": arah,
            "geofon": i + 1,
            "posisi_geofon_m": x_geofon[i],
            "jarak_dari_sumber_m": jarak[i],
            "first_arrival_s": waktu_observasi[i],
            "cabang_teori": indeks_cabang[i] + 1
        })

df = pd.DataFrame(data)

print("\nCONTOH DATA SIMULASI")
print(df.head(10).round(6).to_string(index=False))

# Simpan data simulasi ke CSV
df.to_csv("data_refraksi_simulasi.csv", index=False)
print("\nData tersimpan: data_refraksi_simulasi.csv")

# ------------------------------------------------------------
# 4. REGRESI LINIER TIME-DISTANCE
# ------------------------------------------------------------

# t = intercept + slope * x
# V = 1 / slope
# Hanya gunakan data yang berada pada cabang first arrival
# yang sama menurut model simulasi.

hasil = []

for cabang_id in [1, 2, 3]:
    subset = df[df["cabang_teori"] == cabang_id]

    if len(subset) < 3:
        print(f"Peringatan: cabang {cabang_id} "
              "memiliki data terlalu sedikit.")
        continue

    regresi = linregress(
        subset["jarak_dari_sumber_m"],
        subset["first_arrival_s"]
    )

    slope = regresi.slope
    intercept = regresi.intercept

    if slope <= 0:
        raise ValueError("Slope tidak positif; periksa data.")

    kecepatan = 1 / slope

    hasil.append({
        "cabang": cabang_id,
        "jumlah_data": len(subset),
        "slope_s_per_m": slope,
        "intercept_s": intercept,
        "kecepatan_m_per_s": kecepatan,
        "R2": regresi.rvalue**2
    })

hasil_df = pd.DataFrame(hasil)

print("\nHASIL REGRESI TIME-DISTANCE")
print(hasil_df.round(5).to_string(index=False))

# ------------------------------------------------------------
# 5. ESTIMASI KEDALAMAN DENGAN INTERCEPT TIME
# ------------------------------------------------------------

# Kecepatan dari hasil regresi
# Cabang 1: lapisan pertama
# Cabang 2: lapisan kedua
# Cabang 3: lapisan ketiga

if not all(c in hasil_df["cabang"].values for c in [1, 2, 3]):
    raise ValueError(
        "Salah satu cabang tidak terdeteksi. "
        "Periksa rentang jarak atau parameter model."
    )

def ambil_hasil(cabang, kolom):
    return hasil_df.loc[
        hasil_df["cabang"] == cabang, kolom
    ].iloc[0]

V1_est = ambil_hasil(1, "kecepatan_m_per_s")
V2_est = ambil_hasil(2, "kecepatan_m_per_s")
V3_est = ambil_hasil(3, "kecepatan_m_per_s")

ti2_est = ambil_hasil(2, "intercept_s")
ti3_est = ambil_hasil(3, "intercept_s")

if not (0 < V1_est < V2_est < V3_est):
    raise ValueError(
        "Kecepatan tidak memenuhi V1 < V2 < V3. "
        "Periksa pemilihan cabang dan kualitas data."
    )

# Kedalaman batas lapisan 1-2
h1_est = (
    ti2_est * V1_est * V2_est
    / (2 * np.sqrt(V2_est**2 - V1_est**2))
)

# Kontribusi lapisan pertama pada intercept lapisan ketiga
ti3_kontribusi_h1 = (
    2 * h1_est * np.sqrt(V3_est**2 - V1_est**2)
    / (V1_est * V3_est)
)

# Kedalaman/ketebalan lapisan 2
penyebut_h2 = (
    2 * np.sqrt(V3_est**2 - V2_est**2)
    / (V2_est * V3_est)
)

h2_est = (ti3_est - ti3_kontribusi_h1) / penyebut_h2

if h1_est <= 0 or h2_est <= 0:
    raise ValueError(
        "Kedalaman tidak positif. Model tidak konsisten "
        "dengan waktu intersep hasil regresi."
    )

kedalaman_batas_12 = h1_est
kedalaman_batas_23 = h1_est + h2_est

ringkasan = pd.DataFrame({
    "Parameter": [
        "Kecepatan lapisan 1",
        "Kecepatan lapisan 2",
        "Kecepatan lapisan 3",
        "Waktu intersep lapisan 2",
        "Waktu intersep lapisan 3",
        "Ketebalan lapisan 1",
        "Ketebalan lapisan 2",
        "Kedalaman batas 1-2",
        "Kedalaman batas 2-3"
    ],
    "Nilai": [
        V1_est, V2_est, V3_est,
        ti2_est, ti3_est,
        h1_est, h2_est,
        kedalaman_batas_12,
        kedalaman_batas_23
    ],
    "Satuan": [
        "m/s", "m/s", "m/s", "s", "s",
        "m", "m", "m", "m"
    ]
})

print("\nRINGKASAN HASIL ESTIMASI")
print(ringkasan.round(4).to_string(index=False))

# ------------------------------------------------------------
# 6. GRAFIK TIME-DISTANCE CURVE
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(10, 6))

warna = {"Forward": "tab:blue", "Reverse": "tab:orange"}

for arah in ["Forward", "Reverse"]:
    sub = df[df["arah_shot"] == arah].sort_values(
        "jarak_dari_sumber_m"
    )
    ax.scatter(
        sub["jarak_dari_sumber_m"],
        sub["first_arrival_s"] * 1000,
        label=f"First arrival {arah.lower()}",
        color=warna[arah],
        s=30
    )

# Tambahkan garis regresi setiap cabang
for _, baris in hasil_df.iterrows():
    cabang_id = int(baris["cabang"])
    sub = df[df["cabang_teori"] == cabang_id]

    x_fit = np.linspace(
        sub["jarak_dari_sumber_m"].min(),
        sub["jarak_dari_sumber_m"].max(),
        100
    )

    t_fit = (
        baris["intercept_s"]
        + baris["slope_s_per_m"] * x_fit
    )

    ax.plot(
        x_fit, t_fit * 1000,
        linestyle="--",
        label=f"Regresi cabang {cabang_id}"
    )

ax.set_title("Time-Distance Curve Seismik Refraksi")
ax.set_xlabel("Jarak dari sumber (m)")
ax.set_ylabel("Waktu tiba pertama (ms)")
ax.grid(True, alpha=0.3)
ax.legend()
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 7. PENAMPANG MODEL BAWAH PERMUKAAN 2D
# ------------------------------------------------------------

x_model = np.linspace(0, panjang_lintasan, 250)

# Batas lapisan horizontal berdasarkan estimasi
z12 = h1_est
z23 = h1_est + h2_est
zmax = z23 + 8  # tambahan kedalaman untuk visualisasi

fig, ax = plt.subplots(figsize=(12, 5))

ax.fill_between(
    x_model, 0, z12,
    color="tan", alpha=0.85,
    label=f"Lapisan 1: {V1_est:.0f} m/s"
)
ax.fill_between(
    x_model, z12, z23,
    color="goldenrod", alpha=0.65,
    label=f"Lapisan 2: {V2_est:.0f} m/s"
)
ax.fill_between(
    x_model, z23, zmax,
    color="slategray", alpha=0.65,
    label=f"Lapisan 3: {V3_est:.0f} m/s"
)

ax.axhline(z12, color="black", linewidth=1)
ax.axhline(z23, color="black", linewidth=1)

ax.set_xlim(0, panjang_lintasan)
ax.set_ylim(zmax, 0)
ax.set_xlabel("Jarak sepanjang lintasan (m)")
ax.set_ylabel("Kedalaman (m)")
ax.set_title("Model Bawah Permukaan 2D - Intercept Time")
ax.legend(loc="lower right")
ax.grid(True, alpha=0.2)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 8. INTERPRETASI AWAL KECEPATAN
# ------------------------------------------------------------

def interpretasi_litologi(V):
    if V < 600:
        return "Tanah lepas atau material lapuk; perlu verifikasi"
    elif V < 1500:
        return "Tanah padat atau sedimen; perlu verifikasi"
    elif V < 3000:
        return "Material terkonsolidasi atau batuan lapuk"
    else:
        return "Batuan lebih kompak mungkin; perlu verifikasi"

print("\nINTERPRETASI AWAL")
for i, V in enumerate([V1_est, V2_est, V3_est], start=1):
    print(
        f"Lapisan {i}: Vp = {V:.1f} m/s -> "
        f"{interpretasi_litologi(V)}"
    )

print("\nCatatan: litologi tidak dapat ditentukan hanya "
      "dari kecepatan. Cocokkan dengan data geologi setempat.")
```

### ***PROMPT 3 : PENGOLAHAN DATA METODE MAGNETIK***

#### ***Struktur Prompt***

Judul: Analisis Data Geomagnetik untuk Identifikasi Struktur Geologi Bawah Permukaan

Bertindaklah sebagai ahli geofisika terapan dan dosen pembimbing yang memiliki keahlian dalam metode magnetik. Lakukan pengolahan dan analisis data geomagnetik untuk mengidentifikasi variasi medan magnet bumi, memisahkan anomali regional dan residual, serta menginterpretasikan struktur geologi bawah permukaan berdasarkan distribusi anomali magnetik. Gunakan metode koreksi variasi harian, koreksi IGRF, dan transformasi data magnetik yang sesuai. Sajikan tahapan pengolahan data secara sistematis, perhitungan matematis, kode Python, peta kontur, serta interpretasi geologi berdasarkan hasil analisis.

#### ***R — Role (Peran)***

Berperan sebagai ahli geofisika terapan dan dosen pembimbing yang memiliki keahlian dalam metode geomagnetik, pengolahan data medan magnet bumi, analisis anomali magnetik, pemisahan anomali regional dan residual, serta interpretasi struktur geologi.

#### ***T — Task (Tugas)***

Lakukan pengolahan dan analisis data geomagnetik untuk:
1. Melakukan koreksi variasi harian terhadap data medan magnet hasil pengukuran.
2. Melakukan koreksi IGRF untuk memperoleh anomali magnetik.
3. Membuat peta anomali magnetik total.
4. Memisahkan anomali regional dan residual menggunakan metode yang sesuai.
5. Melakukan transformasi reduksi ke kutub (Reduction to the Pole) jika kondisi data memungkinkan.
6. Mengidentifikasi pola anomali magnetik yang berkaitan dengan struktur geologi bawah permukaan.
7. Menginterpretasikan kemungkinan keberadaan sesar, intrusi batuan, dan kontak litologi berdasarkan pola anomali.

#### ***C — Context (Konteks)***

Metode geomagnetik merupakan metode geofisika yang memanfaatkan variasi medan magnet bumi akibat perbedaan suseptibilitas magnetik batuan. Setiap jenis batuan memiliki karakteristik kemagnetan yang berbeda sehingga dapat menghasilkan anomali magnetik yang bervariasi.

Pengolahan data geomagnetik dilakukan untuk menghilangkan pengaruh medan magnet utama bumi dan variasi harian sehingga diperoleh anomali magnetik yang berkaitan dengan kondisi geologi bawah permukaan. Hasilnya dapat digunakan untuk membantu mengidentifikasi struktur geologi, kontak batuan, dan tubuh batuan yang memiliki sifat magnetik berbeda.

#### ***I — Input (Masukan)***

Data yang digunakan dalam analisis meliputi:
- Data medan magnet total hasil pengukuran.
- Koordinat geografis setiap titik pengukuran.
- Waktu pengukuran pada setiap titik.
- Data variasi harian dari stasiun magnetik dasar jika tersedia.
- Nilai IGRF sesuai lokasi dan waktu pengukuran.
- Ketinggian titik pengukuran jika tersedia.
- Data geologi regional sebagai informasi pendukung.

Apabila data lapangan belum tersedia, gunakan data simulasi yang diberi label dengan jelas dan jangan menyajikannya sebagai data pengukuran sebenarnya.

#### ***K — Constraints (Batasan)***

1. Gunakan koreksi variasi harian dan IGRF sesuai dengan ketersediaan data.
2. Jelaskan setiap tahapan koreksi yang dilakukan.
3. Gunakan metode pemisahan anomali regional dan residual yang sesuai dengan karakteristik data.
4. Jangan menginterpretasikan setiap anomali sebagai sesar atau intrusi tanpa bukti pendukung.
5. Pertimbangkan inklinasi dan deklinasi magnetik dalam penerapan reduksi ke kutub.
6. Gunakan Python dengan pustaka NumPy, Pandas, Matplotlib, dan SciPy jika diperlukan.
7. Bedakan anomali hasil pengolahan dengan interpretasi geologi.

#### ***O — Output (Keluaran)***

Hasil analisis harus mencakup:

1. Pengolahan Data: Penjelasan tahapan koreksi dan pengolahan data geomagnetik.
2. Koreksi Magnetik: Tabel hasil koreksi variasi harian dan IGRF.
3. Peta Anomali Magnetik: Peta kontur anomali magnetik total yang dilengkapi skala warna, koordinat, dan satuan.
4. Pemisahan Anomali: Peta anomali regional dan residual berdasarkan metode yang digunakan.
5. Transformasi Data: Hasil reduksi ke kutub jika sesuai dengan kondisi data.
6. Kode Python: Kode lengkap untuk membaca data, melakukan koreksi, memisahkan anomali, dan membuat peta kontur.
7. Interpretasi Geologi: Penjelasan mengenai pola anomali magnetik yang berpotensi berhubungan dengan sesar, intrusi, atau perbedaan litologi.

#### ***Q — Quality Criteria (Kriteria Kualitas)***

- Koreksi variasi harian dan IGRF harus dilakukan secara tepat berdasarkan data yang tersedia.
- Peta anomali harus memiliki judul, skala warna, koordinat, dan satuan yang jelas.
- Metode pemisahan anomali harus dijelaskan dan memiliki dasar yang sesuai.
- Interpretasi geologi harus mempertimbangkan kemungkinan sumber anomali yang berbeda.
- Kode Python harus dapat dijalankan dan menghasilkan keluaran yang sesuai.
- Hasil analisis disajikan secara sistematis dan dapat digunakan sebagai bahan laporan praktikum mahasiswa S1 Geofisika.

#### ***V — Verification (Verifikasi)***

Sebelum memberikan hasil akhir, lakukan pemeriksaan terhadap:
1. Konsistensi satuan medan magnet dan koordinat pengukuran.
2. Ketepatan koreksi variasi harian dan IGRF.
3. Kesesuaian peta anomali dengan data hasil pengukuran.
4. Konsistensi hasil pemisahan anomali regional dan residual.
5. Kewajaran interpretasi struktur geologi berdasarkan pola anomali magnetik.

### ***PROMPT 4 : PENGOLAHAN DATA METODE GRAVITASI***

#### ***Struktur Prompt***

Judul: Analisis Data Gravitasi untuk Identifikasi Struktur Geologi dan Kontras Densitas Bawah Permukaan

Bertindaklah sebagai ahli geofisika terapan dan dosen pembimbing yang memiliki keahlian dalam metode gravitasi. Lakukan pengolahan dan analisis data gravitasi untuk mengidentifikasi variasi medan gravitasi bumi, menghitung anomali Bouguer, memisahkan anomali regional dan residual, serta menginterpretasikan distribusi densitas dan struktur geologi bawah permukaan. Gunakan koreksi gravitasi yang sesuai, seperti koreksi lintang, udara bebas, Bouguer, dan medan. Sajikan tahapan pengolahan data secara sistematis, perhitungan matematis, kode Python, peta kontur, serta interpretasi geologi berdasarkan hasil analisis.

#### ***R — Role (Peran)***

Berperan sebagai ahli geofisika terapan dan dosen pembimbing yang memiliki keahlian dalam metode gravitasi, pengolahan data gravitasi, koreksi medan gravitasi, analisis anomali Bouguer, pemodelan densitas, serta interpretasi struktur geologi bawah permukaan.

#### ***T — Task (Tugas)***

Lakukan pengolahan dan analisis data gravitasi untuk:

#### ***C — Context (Konteks)***

Perbedaan densitas batuan dapat menghasilkan anomali gravitasi yang digunakan untuk memperkirakan keberadaan struktur geologi, cekungan sedimen, intrusi batuan, maupun zona dengan kontras densitas tertentu.

Pengolahan data gravitasi dilakukan melalui beberapa koreksi untuk menghilangkan pengaruh faktor nongeologi, seperti lintang, elevasi, dan topografi. Hasil pengolahan berupa anomali gravitasi yang dapat digunakan untuk membantu interpretasi kondisi bawah permukaan.

#### ***I — Input (Masukan)***

Data yang digunakan dalam analisis meliputi:
- Data gravitasi hasil pengukuran lapangan.
- Koordinat geografis setiap titik pengukuran.
- Elevasi setiap titik pengukuran.
- Nilai gravitasi teoritis atau gravitasi normal.
- Data topografi untuk koreksi medan jika tersedia.
- Nilai densitas batuan rata-rata untuk koreksi Bouguer.
- Data geologi regional sebagai informasi pendukung.

Apabila data lapangan belum tersedia, gunakan data simulasi yang diberi label dengan jelas dan jangan menyajikannya sebagai data pengukuran sebenarnya.

#### ***K — Constraints (Batasan)***

1. Gunakan persamaan koreksi gravitasi yang sesuai dengan data yang tersedia.
2. Tampilkan tahapan perhitungan koreksi gravitasi secara sistematis.
3. Jelaskan asumsi nilai densitas yang digunakan dalam koreksi Bouguer.
4. Gunakan metode pemisahan anomali regional dan residual yang sesuai dengan karakteristik data.
5. Jangan menginterpretasikan anomali gravitasi secara langsung sebagai jenis batuan tanpa mempertimbangkan informasi geologi.
6. Gunakan Python dengan pustaka NumPy, Pandas, Matplotlib, dan SciPy jika diperlukan.
7. Bedakan hasil pengukuran, hasil koreksi, hasil pemodelan, dan interpretasi geologi.

#### ***O — Output (Keluaran)***

Hasil analisis harus mencakup:

1. Pengolahan Data: Penjelasan tahapan pengolahan data gravitasi dari data mentah hingga diperoleh anomali Bouguer.
2. Koreksi Gravitasi: Tabel hasil koreksi lintang, udara bebas, Bouguer, dan medan jika tersedia.
3. Anomali Bouguer: Tabel hasil perhitungan anomali Bouguer lengkap pada setiap titik pengukuran.
4. Peta Anomali: Peta kontur anomali Bouguer lengkap yang menunjukkan distribusi anomali gravitasi.
5. Pemisahan Anomali: Peta anomali regional dan residual menggunakan metode yang dipilih.
6. Kode Python: Kode lengkap untuk membaca data, melakukan koreksi gravitasi, menghitung anomali, dan membuat peta kontur.
7. Interpretasi Geologi: Penjelasan mengenai kemungkinan variasi densitas, struktur geologi, dan batas litologi berdasarkan pola anomali gravitasi.

#### ***Q — Quality Criteria (Kriteria Kualitas)***

- Perhitungan koreksi gravitasi harus sesuai dengan persamaan yang digunakan.
- Seluruh satuan harus konsisten, terutama mGal, meter, dan g/cm³.

1. Konsistensi satuan gravitasi, elevasi, densitas, dan koordinat.

### ***PROMPT 5 : PENGOLAHAN DATA MAGNETOTELURIK (MT)***

#### ***Struktur Prompt***

Judul: Analisis Data Magnetotelurik untuk Identifikasi Struktur Resistivitas Bawah Permukaan dan Eksplorasi Panas Bumi

Bertindaklah sebagai ahli geofisika terapan dan dosen pembimbing yang memiliki keahlian dalam metode magnetotelurik (MT). Lakukan pengolahan dan analisis data magnetotelurik untuk mengidentifikasi distribusi resistivitas bawah permukaan berdasarkan variasi medan listrik dan medan magnet alami bumi. Gunakan analisis impedansi, kurva resistivitas semu, fase, serta pemodelan inversi 1D atau 2D untuk memperkirakan struktur resistivitas bawah permukaan. Sajikan tahapan pengolahan data secara sistematis, persamaan matematis, kode Python jika memungkinkan, grafik, serta interpretasi geologi yang berkaitan dengan sistem panas bumi.

#### ***R — Role (Peran)***

Berperan sebagai ahli geofisika terapan dan dosen pembimbing yang memiliki keahlian dalam metode magnetotelurik, analisis medan elektromagnetik alami, pengolahan data MT, pemodelan resistivitas, dan interpretasi sistem panas bumi.

#### ***T — Task (Tugas)***

Lakukan pengolahan dan analisis data magnetotelurik untuk:
1. Mengolah data medan listrik dan medan magnet alami bumi.
2. Menghitung impedansi elektromagnetik berdasarkan komponen medan listrik dan medan magnet.
3. Menghitung resistivitas semu dan fase pada setiap frekuensi.
4. Membuat kurva resistivitas semu dan fase terhadap frekuensi atau periode.
5. Melakukan pemodelan inversi 1D atau 2D untuk memperoleh distribusi resistivitas bawah permukaan.
6. Mengidentifikasi zona konduktif dan resistif berdasarkan hasil inversi.
7. Menginterpretasikan kemungkinan keberadaan zona alterasi, reservoir, dan struktur geologi yang berhubungan dengan sistem panas bumi.

#### ***C — Context (Konteks)***

Metode magnetotelurik merupakan metode geofisika elektromagnetik pasif yang memanfaatkan variasi alami medan listrik dan medan magnet bumi untuk menyelidiki struktur resistivitas bawah permukaan. Metode ini memiliki kemampuan penetrasi yang relatif dalam, bergantung pada periode sinyal dan kondisi resistivitas bawah permukaan.

Dalam eksplorasi panas bumi, metode MT dapat digunakan untuk mengidentifikasi zona konduktif yang mungkin berkaitan dengan batuan teralterasi hidrotermal, serta zona resistif yang berpotensi berhubungan dengan reservoir atau batuan dasar. Interpretasi harus mempertimbangkan informasi geologi dan geokimia karena anomali resistivitas tidak selalu menunjukkan keberadaan sistem panas bumi.

#### ***I — Input (Masukan)***

Data yang digunakan dalam analisis meliputi:
- Data medan listrik horizontal (Ex dan Ey).
- Data medan magnet horizontal (Hx dan Hy).
- Data medan magnet vertikal (Hz) jika tersedia.
- Data frekuensi atau periode pengukuran.
- Data impedansi elektromagnetik jika telah tersedia.
- Koordinat lokasi pengukuran MT.
- Informasi geologi dan geokimia daerah penelitian jika tersedia.

Apabila data lapangan belum tersedia, gunakan data simulasi yang diberi label dengan jelas dan jangan menyajikannya sebagai data pengukuran sebenarnya.

#### ***K — Constraints (Batasan)***

1. Gunakan persamaan impedansi elektromagnetik yang sesuai dengan komponen medan listrik dan medan magnet.
2. Tampilkan tahapan perhitungan resistivitas semu dan fase secara jelas.
3. Gunakan metode inversi yang sesuai dengan dimensi dan kualitas data.
4. Jelaskan hubungan antara frekuensi, periode, dan kedalaman penetrasi elektromagnetik.
5. Jangan menginterpretasikan zona konduktif secara langsung sebagai reservoir panas bumi tanpa bukti pendukung.
6. Gunakan Python dengan pustaka NumPy, Pandas, Matplotlib, dan SciPy jika diperlukan.
7. Bedakan hasil pengukuran, hasil inversi, dan interpretasi geologi.

#### ***O — Output (Keluaran)***

Hasil analisis harus mencakup:

1. Pengolahan Data: Penjelasan tahapan pengolahan data MT dari data medan elektromagnetik hingga diperoleh parameter impedansi.
2. Kurva Resistivitas Semu: Grafik resistivitas semu terhadap frekuensi atau periode.
3. Kurva Fase: Grafik fase impedansi terhadap frekuensi atau periode.
4. Hasil Inversi: Model resistivitas bawah permukaan berdasarkan hasil inversi 1D atau 2D.
5. Kode Python: Kode untuk membaca data, menghitung parameter MT, dan membuat grafik jika data serta pustaka yang tersedia memungkinkan.
6. Penampang Resistivitas: Model 2D yang memperlihatkan distribusi resistivitas terhadap kedalaman dan jarak.
7. Interpretasi Geologi: Penjelasan mengenai zona konduktif, zona resistif, dan kemungkinan struktur yang berkaitan dengan sistem panas bumi.

#### ***Q — Quality Criteria (Kriteria Kualitas)***

- Perhitungan impedansi, resistivitas semu, dan fase harus sesuai dengan persamaan elektromagnetik.
- Grafik harus memiliki judul, label sumbu, satuan, dan legenda yang jelas.
- Hasil inversi harus disertai informasi mengenai kecocokan model terhadap data.
- Interpretasi harus mempertimbangkan ambiguitas resistivitas dan informasi geologi pendukung.
- Kode Python harus dapat dijalankan untuk bagian pengolahan yang diimplementasikan.
- Hasil analisis disajikan secara sistematis dan dapat digunakan sebagai bahan laporan praktikum mahasiswa S1 Geofisika.

#### ***V — Verification (Verifikasi)***

Sebelum memberikan hasil akhir, lakukan pemeriksaan terhadap:
1. Konsistensi satuan medan listrik, medan magnet, frekuensi, dan resistivitas.
2. Ketepatan perhitungan impedansi, resistivitas semu, dan fase.
3. Kesesuaian kurva hasil pengolahan dengan data pengukuran.
4. Kesesuaian hasil inversi dengan respons elektromagnetik yang terukur.
